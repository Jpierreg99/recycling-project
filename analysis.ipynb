```python
poblacion = poblacion.drop_duplicates()

residuos = residuos.drop_duplicates()

reciclaje = reciclaje.drop_duplicates()

print("Duplicates removed")
```

```python
# Population category

poblacion['POPULATION_CATEGORY'] = poblacion['POBLACION'].apply(
    lambda x: 'High' if x > 50000 else 'Low'
)

print(poblacion[['POBLACION', 'POPULATION_CATEGORY']].head())
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

poblacion['SEXO_ENCODED'] = encoder.fit_transform(poblacion['SEXO'])

print(poblacion[['SEXO', 'SEXO_ENCODED']].head())
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

poblacion['POBLACION_NORMALIZED'] = scaler.fit_transform(
    poblacion[['POBLACION']]
)

print(
    poblacion[['POBLACION', 'POBLACION_NORMALIZED']].head()
)
```

```python
clustering_data = poblacion[[
    'POBLACION_NORMALIZED',
    'SEXO_ENCODED'
]]

print(clustering_data.head())
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(
    n_clusters=3,
    random_state=42
)

kmeans.fit(clustering_data)

poblacion['CLUSTER'] = kmeans.labels_

print(
    poblacion[['POBLACION', 'CLUSTER']].head()
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))

plt.scatter(
    poblacion['POBLACION_NORMALIZED'],
    poblacion['SEXO_ENCODED'],
    c=poblacion['CLUSTER']
)

plt.title('K-Means Clustering')
plt.xlabel('Normalized Population')
plt.ylabel('Encoded Gender')

plt.show()
```

```python
plt.figure(figsize=(12,6))

plt.scatter(
    poblacion.index,
    poblacion['POBLACION_NORMALIZED'],
    c=poblacion['CLUSTER'],
    s=20
)

plt.title('Clusters by Population')
plt.xlabel('Records')
plt.ylabel('Normalized Population')

plt.show()
```
